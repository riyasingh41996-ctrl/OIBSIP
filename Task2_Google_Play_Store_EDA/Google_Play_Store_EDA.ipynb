```python
df['Rating'] = df['Rating'].fillna(df['Rating'].median())
```

```python
df['Type'] = df['Type'].fillna(df['Type'].mode()[0])
```

```python
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
```

```python
df['Reviews'].dtype
```

```python
df['Installs'] = pd.to_numeric(
    df['Installs'].astype(str).str.replace(',', '').str.replace('+', ''),
    errors='coerce'
)
```

```python
df['Installs'].dtype
```

```python
df['Price'].unique()[:30]
```

```python
df['Price'] = pd.to_numeric(
    df['Price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False),
    errors='coerce'
)
```

```python
df['Price'].dtype
```

```python
df['Price'].isnull().sum()
```

```python
df['Price'] = df['Price'].fillna(df['Price'].median())
```

```python
df['Price'].isnull().sum()
```

```python
df['Installs'] = pd.to_numeric(
    df['Installs'].astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('+', '', regex=False),
    errors='coerce'
)
```

```python
df['Installs'].dtype
```

```python
df['Installs'].isnull().sum()
```

```python
df['Installs'].isnull().sum()
```

```python
df['Installs'] = df['Installs'].fillna(df['Installs'].median())
```

```python
df[['App', 'Rating']].sort_values(by='Rating', ascending=False).head(10)
```

```python
df['Category'].value_counts().head(10).plot(kind='bar', figsize=(10,5))
plt.title('Top 10 App Categories')
plt.xlabel('Category')
plt.ylabel('Number of Apps')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Rating'], bins=10)
plt.title('Distribution of App Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Apps')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Rating'], df['Installs'], alpha=0.3)
plt.title('Rating vs Number of Installs')
plt.xlabel('Rating')
plt.ylabel('Installs')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Rating'], df['Installs'], alpha=0.3)
plt.yscale('log')
plt.title('Rating vs Number of Installs')
plt.xlabel('Rating')
plt.ylabel('Installs (Log Scale)')
plt.show()
```

```python
df['Type'].value_counts().plot(kind='bar', figsize=(6,4))

plt.title('Free vs Paid Apps')
plt.xlabel('App Type')
plt.ylabel('Number of Apps')
plt.show()
```

```python
top10 = df[['App', 'Installs']].sort_values(
    by='Installs', ascending=False
).head(10)

top10
```

```python
plt.figure(figsize=(10,5))

plt.bar(top10['App'], top10['Installs'])

plt.xticks(rotation=90)
plt.title('Top 10 Apps by Installs')
plt.xlabel('App')
plt.ylabel('Number of Installs')

plt.show()
```

```python
plt.figure(figsize=(8,5))

df.groupby('Category')['Installs'].sum().sort_values(ascending=False).head(10).plot(kind='bar')

plt.title('Top Categories by Total Installs')
plt.xlabel('Category')
plt.ylabel('Total Installs')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Rating'], df['Installs'], alpha=0.3)

plt.title('Rating vs Number of Installs')
plt.xlabel('Rating')
plt.ylabel('Number of Installs')

plt.show()
```
